```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", context="notebook")

DATA = Path("../data/heart_clean.csv")      # notebook lives in notebooks/
FIG_DIR = Path("../screenshots/eda")
FIG_DIR.mkdir(parents=True, exist_ok=True)

def save(name):
    plt.tight_layout()
    plt.savefig(FIG_DIR / f"{name}.png", dpi=150, bbox_inches="tight")

df = pd.read_csv(DATA)
print("Shape:", df.shape)
df.head()
```

## 1. Data dictionary

| Column | Meaning | Type |
|---|---|---|
| age | Age in years | numeric |
| sex | 1 = male, 0 = female | binary |
| cp | Chest pain type (1-4) | categorical |
| trestbps | Resting blood pressure (mm Hg) | numeric |
| chol | Serum cholesterol (mg/dl) | numeric |
| fbs | Fasting blood sugar > 120 mg/dl (1 = yes) | binary |
| restecg | Resting ECG result (0-2) | categorical |
| thalach | Maximum heart rate achieved | numeric |
| exang | Exercise-induced angina (1 = yes) | binary |
| oldpeak | ST depression induced by exercise | numeric |
| slope | Slope of peak exercise ST segment (1-3) | categorical |
| ca | Number of major vessels coloured by fluoroscopy (0-3) | categorical/count |
| thal | Thalassemia: 3 = normal, 6 = fixed defect, 7 = reversible defect | categorical |
| target | 0 = no disease, 1 = disease present | **label** |

```python
df.info()
print()
print("Duplicate rows:", df.duplicated().sum())
print()
print("Missing values per column:")
print(df.isna().sum()[df.isna().sum() > 0])
```

## 2. Class balance
Is the target roughly balanced? This decides whether accuracy alone is a fair metric.

```python
counts = df["target"].value_counts().sort_index()
pct = (counts / len(df) * 100).round(1)
print(pd.DataFrame({"count": counts, "percent": pct}))

fig, ax = plt.subplots(figsize=(5, 4))
sns.countplot(data=df, x="target", ax=ax)
ax.set_xticklabels(["No disease (0)", "Disease (1)"])
ax.set_title("Class balance")
for p in ax.patches:
    ax.annotate(int(p.get_height()), (p.get_x() + p.get_width() / 2, p.get_height()),
                ha="center", va="bottom")
save("class_balance")
plt.show()
```

**Takeaway:** The dataset is fairly balanced (164 patients without disease, 139 with, i.e. 54.1% and 45.9%). Accuracy is therefore a fair summary metric and no resampling is needed. Precision, recall and ROC-AUC are still reported, because in a heart-disease screening setting missing a sick patient (low recall) is more costly than a false alarm.

## 3. Numeric feature distributions

```python
num_cols = ["age", "trestbps", "chol", "thalach", "oldpeak"]

fig, axes = plt.subplots(2, 3, figsize=(14, 7))
for ax, col in zip(axes.ravel(), num_cols):
    sns.histplot(data=df, x=col, kde=True, ax=ax)
    ax.set_title(col)
axes.ravel()[-1].axis("off")
fig.suptitle("Distributions of numeric features", y=1.02, fontsize=14)
save("numeric_histograms")
plt.show()
df[["age","trestbps","chol","thalach","oldpeak"]].skew()
```

**Takeaway:** oldpeak is strongly right-skewed with many zeros, chol is mildly right-skewed with a few very high values, and trestbps is slightly right-skewed. age is roughly symmetric and thalach slightly left-skewed. No impossible values were found, so no rows were removed. Standardisation is used for the numeric features, and medians are used for imputation because they are robust to outliers.

```python
# Same features split by target - which ones separate the classes?
fig, axes = plt.subplots(2, 3, figsize=(14, 7))
for ax, col in zip(axes.ravel(), num_cols):
    sns.boxplot(data=df, x="target", y=col, ax=ax)
    ax.set_title(f"{col} by target")
axes.ravel()[-1].axis("off")
save("numeric_by_target")
plt.show()

for c in ["sex", "cp", "fbs", "exang", "slope", "ca", "thal"]:
    print(df.groupby(c)["target"].mean().round(2), "\n")
```

## 4. Correlation heatmap

```python
corr = df.corr(numeric_only=True)

plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0, square=True,
            cbar_kws={"shrink": 0.8})
plt.title("Correlation heatmap")
save("correlation_heatmap")
plt.show()

print("Correlation with target (sorted):")
print(corr["target"].drop("target").sort_values(ascending=False).round(2))
```

**Takeaway:** The features most positively correlated with the target are thal (0.53), ca (0.46), exang (0.43), oldpeak (0.42) and cp (0.41), followed by slope (0.34), sex (0.28) and age (0.22). thalach (maximum heart rate) is the only clearly negative one (-0.42): patients who reach a lower maximum heart rate are more likely to have disease. fbs (0.03) and chol (0.09) show almost no linear relationship with the target.

## 5. Categorical features vs target

```python
cat_cols = ["sex", "cp", "fbs", "restecg", "exang", "slope", "ca", "thal"]

fig, axes = plt.subplots(2, 4, figsize=(18, 8))
for ax, col in zip(axes.ravel(), cat_cols):
    sns.countplot(data=df, x=col, hue="target", ax=ax)
    ax.set_title(f"{col} vs target")
    ax.legend(title="target", loc="upper right")
save("categorical_vs_target")
plt.show()
```

**Takeaway:** thal, ca, exang and cp separate the classes most clearly, and sex and slope less so; fbs shows almost no difference in disease rate between its two values.

```python
# Disease rate by chest pain type and by age band
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

cp_rate = df.groupby("cp")["target"].mean() * 100
sns.barplot(x=cp_rate.index.astype(int), y=cp_rate.values, ax=axes[0], color="#4C78A8")
axes[0].set(title="Disease rate by chest pain type", xlabel="Chest pain type (cp)", ylabel="% with disease")

bands = pd.cut(df["age"], bins=[28, 40, 50, 60, 80], labels=["<=40", "41-50", "51-60", "61+"])
band_rate = df.groupby(bands, observed=True)["target"].mean() * 100
sns.barplot(x=band_rate.index.astype(str), y=band_rate.values, ax=axes[1], color="#E45756")
axes[1].set(title="Disease rate by age band", xlabel="Age band", ylabel="% with disease")

save("disease_rate_by_chest_pain_and_age")
plt.show()
```

```python
# Age vs maximum heart rate, coloured by disease
plot_df = df.assign(Disease=df["target"].map({0: "No disease", 1: "Disease"}))
plt.figure(figsize=(7.5, 5))
sns.scatterplot(data=plot_df, x="age", y="thalach", hue="Disease", alpha=0.8,
                palette={"No disease": "#4C78A8", "Disease": "#E45756"})
plt.title("Age vs maximum heart rate")
plt.xlabel("Age (years)")
plt.ylabel("Maximum heart rate (thalach)")
save("age_vs_max_heart_rate")
plt.show()
```
